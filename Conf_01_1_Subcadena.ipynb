```python
subcadena1 = "LatLng("
subcadena2 = "),icon:"
subcadena3 = "<h1 class=""left"">' + '"
subcadena4 = "' + '</h1>' + '"
```

Ahora calcularemos los índices iniciales y finales de la subcadena que contiene la latitud y la longitud y generaremos una subcadena a partir de los dos valores

```python
ilatlong = cadena.find(subcadena1)+7
flatlong = cadena.find(subcadena2)
cadena_1 = cadena[ilatlong:flatlong]
latlong = cadena_1.split(sep=",")
```
