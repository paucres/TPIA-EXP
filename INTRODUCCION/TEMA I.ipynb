## Entornos de Ejecución en Python

### Objetivos

Al finalizar este apartado serás capaz de:

- Comprender qué es un entorno de ejecución.
- Diferenciar entre Python, un entorno de Python y una distribución como Anaconda.
- Entender por qué una biblioteca puede estar disponible en un entorno y no en otro.
- Conocer cómo Dataiku gestiona sus Code Environments.

Antes de comenzar a programar es importante comprender el concepto de **entorno de ejecución** (*execution environment*). Un entorno de ejecución es el conjunto de recursos que permiten ejecutar un programa en Python: el intérprete del lenguaje, las bibliotecas instaladas, las variables de configuración y las dependencias necesarias para que el código funcione correctamente.

En la práctica, un mismo programa puede comportarse de forma diferente según el entorno en el que se ejecute. Por ejemplo, dos entornos pueden utilizar versiones distintas de Python o disponer de conjuntos diferentes de bibliotecas. Por ello, que un programa funcione correctamente en un ordenador o plataforma no garantiza que lo haga en otro.

Durante esta asignatura trabajaremos utilizando principalmente dos herramientas:

* Visual Studio Code con la Extensión de Jupyter Notebook como IDE de programación habitual.
* Dataiku, una plataforma orientada al análisis de datos y al desarrollo de soluciones de Inteligencia Artificial. Dataiku permite combinar programación visual con el desarrollo de código mediante **recetas de Python** y **cuadernos de Jupyter Notebook**, facilitando la integración entre ambos enfoques. https://www.dataiku.com/

Cada proyecto de Inteligencia Artificial utiliza un **entorno de código (Code Environments)**, que contiene una versión específica de Python y los módulos y bibliotecas disponibles para ejecutar el código. 

En una instalación limpia de Python, el único módulo que se encuentra disponible es PIP. Esto implica que, todas las bibliotecas habituales en Ciencia de Datos e Inteligencia Artificial, como **Pandas**, **Numpy**, **Scipy**, **Matplotlib**, **SciPy** o **Scikit-learn**, no están instaladas por defecto y deben añadirse al entorno antes de poder utilizarlas.

Por este motivo, una parte importante del trabajo consistirá en aprender a identificar el entorno de ejecución que se está utilizando, comprobar qué bibliotecas están disponibles e instalar aquellas dependencias que sean necesarias. Comprender este proceso resulta fundamental para desarrollar aplicaciones reproducibles y evitar errores relacionados con la ausencia o incompatibilidad de módulos.

Para saber que módulos o librerias se encuentran instalados en el entorno utilizamos el siguiente comando desde la consola:

PIP list

Para Instalar una librería utilizamos el comando desde la consola:

PIP install

Por ejemplo: PIP install scikit-learn instala la librería de machine learning sklearn

Es posible instalar más de una librería o módulo a la vez:

PIP install scikit-learn pandas instala sklearn y pandas.

Sin embargo, en la mayoría de los Proyectos se requiere instalar gran cantidad de librerías y modulos por lo que una buena práctica es construir un fichero de requerimientos del entorno de ejecución. Esta filosofía permite garantizar que tanto el entorno de desarrollo como el de despliegue tienen instalado ***las mismas librerías, módulos o paquetes*** y con la **misma versión**.

Para crear el fichero de requerimientos desde la consola utilizamos el comando:

PIP freeze > requirements.txt

Para posteriormente instalar el fichero de requerimientos dentro del entorno utilizamos:

PIP install -r requirements.txt

Una vez instalados podemos utilizar las librerías dentro de nuestro código utilizando el comando _import_.

Una **buena práctica**, en librerías o módulos que se van a utilizar muy frecuente, es importar las librerías utilizando alias. Esto permite que el código sea más eficiente.

```python
import pandas as pd
import numpy as np
import scipy as sp
import sklearn
import matplotlib.pyplot as plt
```

### SciPy

SciPy amplía NumPy incorporando algoritmos científicos.

https://scipy.org/

```python
datos = [5, 7, 8, 6, 9, 10, 8]

describe_datos = sp.stats.describe(datos)

print(describe_datos)
```

La salida del **print** es bastante ilegible así que aprovecharemos para mejorarla dándole un formato más elegante:

```python
print("=== Estadísticos descriptivos ===")
print(f"Número de observaciones : {describe_datos.nobs}")
print(f"Valor mínimo            : {describe_datos.minmax[0]}")
print(f"Valor máximo            : {describe_datos.minmax[1]}")
print(f"Media                   : {describe_datos.mean:.2f}")
print(f"Varianza                : {describe_datos.variance:.2f}")
print(f"Asimetría (Skewness)    : {describe_datos.skewness:.3f}")
print(f"Curtosis                : {describe_datos.kurtosis:.3f}")
```

Cuando solamente vamos a utilizar un módulo específico de una librería no es necesario importar la librería completa. Podemos importar solo ese módulo. De esta forma el código queda mucho más limpio.

```python
from scipy import stats #Esta opción es la recomendada cuando vamos a utilizar solamente ciertos módulos de una librería específica

descriptivas = stats.describe(vector)
print("=== Estadísticos descriptivos ===")
print(f"Número de observaciones : {descriptivas.nobs}")
print(f"Valor mínimo            : {descriptivas.minmax[0]}")
print(f"Valor máximo            : {descriptivas.minmax[1]}")
print(f"Media                   : {descriptivas.mean:.2f}")
print(f"Varianza                : {descriptivas.variance:.2f}")
print(f"Asimetría (Skewness)    : {descriptivas.skewness:.3f}")
print(f"Curtosis                : {descriptivas.kurtosis:.3f}")
```

### Matplotlib

Permite representar datos gráficamente.

```python
# import matplotlib.pyplot as plt

X = np.array([1,2,3,4,5])
y = np.array([1,4,9,16,25])

plt.scatter(X,y)

plt.title("Ejemplo de gráfica")
plt.xlabel("X")
plt.ylabel("y")

plt.show()
```

Matplotlib da muchísima facilidad para crear gráficas complejas. Por ejemplo, podemos incluir varias variables que comparten la misma x en un único gráfico:

```python
plt.style.use('_mpl-gallery')

fig, ax = plt.subplots()

y_times_4 = y * 4
y_power_2 = y ** 2
y_squared_3 = y ** (1/3)

plt.title("Gráfica con Subgráficas")
plt.xlabel("X")
plt.ylabel("y")

ax.plot(X, y_times_4, 'x', markeredgewidth=1.5)
ax.plot(X, y_power_2, linewidth=1.5)
ax.plot(X, y_squared_3, 'o-', linewidth=1.5)

plt.show()
```

### Condicionales

Cuenta con tres bloques: **if, else** y **elif**

El bloque **else** es opcional. El bloque **elif** (else if) permite introducir más bloques **if** en caso de que no se cumpla el anterior.

```python
edad = int(input('Ingresa tu edad: ')) # Hay que forzar el tipo de variable a int. Si no se fuerza, por defecto el input se guarda como un string

rango = np.arange(101)

if edad in rango:
    if edad < 10:
        print("Estás en la etapa de la niñez según la OMS")
    elif edad <=19:
        print("Estás en la adolescencia según la OMS")
    else:
        print("Estás en la etapa adulta según la OMS")
else:
    print('La edad introducida está fuera de rango')
```

### While

Opera exactamente igual que en C.

El cuerpo del bucle debe estar sangrado, de lo contrario, nos aparece un error de sangrado.

```python
i = 1

while i <= 5:
    print(i)
    i += 1
```

### For

Opera distinto a otros lenguajes de programación como C o Pascal.

Permite iterar sobre elementos de cualquier secuencia (listas, cadenas de texto, etc.) en el orden en que aparecen en la secuencia.

```python
frutas = ["Manzana","Pera","Naranja"]

for fruta in frutas:
    print(fruta)
```

### Range

Genera progresiones aritméticas cuando necesitamos iterar dentro de un **for**

```python
for i in range(5):
    print(i)
```

### Break

Es útil para finalizar bucles **for** o **while** anidados

```python
for i in range(10):

    if i == 5:
        break

    print(i)
```

### Continue

Continua la siguiente iteración del ciclo

```python
for i in range(10):

    if i % 2 == 0:
        continue

    print(i)
```

### Ideas clave

- Las estructuras de control determinan el flujo de ejecución del programa.
- Python utiliza la indentación para definir bloques de código.
- Los bucles for recorren secuencias de elementos.
- range() genera secuencias de números enteros.

**Errores frecuentes**

- Olvidar la indentación después de `if`, `for` o `while`.
- Confundir `=` con `==`.
- Crear bucles `while` infinitos al no actualizar la variable de control.

## Excepciones

Una excepción es un error que ocurre durante la ejecución de un programa (tiempo de ejecución), interrumpiendo su flujo normal. En lugar de permitir que el programa falle e imprima un mensaje incomprensible para el usuario, Python permite capturar y manejar estos errores de forma controlada.

### Objetivos

Al finalizar este apartado serás capaz de:

- Conocer la estructura básica de una excepción
- Capturar excepciones específicas
- Conocer los bloques complementarios
- Lanzar excepciones manualmente

### Estructura Básica

Para manejar excepciones se utiliza el bloque try y except:

* try: Contiene el código que podría generar un error.
* except: Contiene el código que se ejecutará únicamente si ocurre una excepción.

```python
try:
    numero = int(input("Introduce un número entero: "))
    resultado = 10 / numero
    print(f"El resultado es: {resultado}")
except:
    print("Ha ocurrido un error al procesar el número.")
```

### Captura de Excepciones Específicas

Es una **buena práctica** indicar qué tipo de excepción concreta esperas capturar, en lugar de usar un except genérico. Esto evita enmascarar errores inesperados.

Las principales excepciones nativas de Python son:

- ZeroDivisionError
- ValueError
- TypeError
- IndexError
- KeyError
- FileNotFoundError

```python
try:
    numero = int("hola")  # Genera ValueError
    resultado = 10 / 0    # Genera ZeroDivisionError
except ZeroDivisionError:
    print("Error: No se puede dividir entre cero.")
except ValueError:
    print("Error: El valor ingresado no se puede convertir a número.")
```

### Bloques Complementarios

Se pueden añadir dos bloques opcionales al control de excepciones:

* else: Se ejecuta solo si **NO ocurrió ningún error** en el bloque try.
* finally: Se ejecuta **SIEMPRE**, haya ocurrido o no una excepción (muy útil para liberar recursos como cerrar archivos o conexiones).

```python
try:
    archivo = open("datos.txt", "r")
    contenido = archivo.read()
except FileNotFoundError:
    print("El archivo no se ha encontrado.")
else:
    print("Archivo leído con éxito.")
finally:
    print("Finalizando operación (limpieza de recursos).")
```

### Lanzar Excepciones Manualmente

Tú mismo puedes forzar la aparición de una excepción cuando se incumpla una regla de negocio o validación de tu programa mediante la palabra clave raise.

```python
def validar_edad(edad):
    if edad < 0:
        raise ValueError("La edad no puede ser un número negativo.")
    return f"Edad válida: {edad}"

try:
    print(validar_edad(-5))
except ValueError as error:
    print(f"Excepción capturada: {error}")
```

### Ideas clave

- Evitan caídas inesperadas: Las excepciones permiten capturar errores en tiempo de ejecución y gestionarlos de forma controlada sin detener el programa.
- Identificación específica: Identificar la excepción concreta (ValueError, IndexError, etc.) es fundamental para aplicar la solución correcta a cada tipo de problema.
- Control total del flujo: else asegura la ejecución cuando no hay errores, mientras que finally se ejecuta siempre para limpiar recursos (como cerrar archivos).
- Control de reglas de negocio: La palabra clave raise permite forzar excepciones propias cuando los datos no cumplen con las condiciones requeridas.
