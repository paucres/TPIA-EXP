# Introducción.

En el ejemplo anterior nos percatamos que cuando la acción que se realiza en un cuaderno depende del usuario, existe la posibilidad que el usuario introduzca un valor que rompa nuestro código.

En ese sentido, el objetivo de este cuaderno es el de capturar los posibles errores para que el código no se rompa.

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
```

```python
def bisiesto(n):

    result = []

    if n % 400 == 0:
        secular = True
        bisiesto = True
    elif n % 100 == 0:
        secular = False
        bisiesto = False
    elif n % 4 == 0:
        secular = False
        bisiesto = True
    else:
        secular = False
        bisiesto = False

    if n % 100 == 0:
        mult100 = True
    else:
        mult100 = False

    if n % 4 ==0:
        mult4 = True
    else:
        mult4 = False

    result.append(n)
    result.append(bisiesto)
    result.append(secular)
    result.append(mult100)
    result.append(mult4)

    return result
```

```python
def diasemana(n):

    a=np.remainder(1+5*np.remainder((n-1),4)+4*np.remainder((n-1),100)+6*np.remainder((n-1),400),7)

    if a == 0:
        dia = "domingo"
    elif a == 1:
        dia = "lunes"
    elif a == 2:
        dia = "martes"
    elif a == 3:
        dia = "miércoles"
    elif a == 4:
        dia = "jueves"
    elif a == 5:
        dia = "viernes"
    else:
        dia = "sábado"

    return dia
```

```python
df = pd.DataFrame()

años = []
dias = []

for i in range(1, 3001):
    bis = bisiesto(i)
    day = diasemana(i)

    años.append(bis)
    dias.append(day)

df = pd.DataFrame(años, columns=['Año', 'Bisiesto', 'Secular', 'Múltiplo_100', 'Múltiplo_4'])
df['Comienza en'] = dias
df = df.reindex(columns=['Año', 'Comienza en', 'Bisiesto', 'Secular', 'Múltiplo_100', 'Múltiplo_4'])
df
```

```python
X = df[['Año', 'Secular', 'Múltiplo_100', 'Múltiplo_4']]
Y = df[['Bisiesto']]

print(X)
print(Y)

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.3, random_state=0)

logisticregression = LogisticRegression()

logisticregression.fit(x_train, y_train.values.ravel())

predictions = logisticregression.predict(x_test)
score = logisticregression.score(x_test, y_test)
params = logisticregression.get_params(deep=True)

print(score)
```
