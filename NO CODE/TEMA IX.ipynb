```python
import pandas as pd
import numpy as np
import streamlit as st
import gradio as gr
import io
from lazypredict.Supervised import LazyClassifier
from sklearn.model_selection import train_test_split
from flaml import AutoML
from sklearn.metrics import f1_score
```

```python
# Ejemplo conceptual: De la intención al código asistido
"""
INTENCIÓN (Vibe Coding Prompt):
"Crea una función en Python utilizando pandas que lea un DataFrame, 
identifique columnas numéricas con asimetría superior a 1.5 y aplique 
una transformación log1p solo a esas columnas, devolviendo el DataFrame 
modificado y la lista de columnas transformadas."
"""

# CÓDIGO GENERADO AUTOMÁTICAMENTE POR EL AGENTE:

def corregir_asimetria(df: pd.DataFrame, umbral: float = 1.5):
    df_transformed = df.copy()
    cols_numericas = df_transformed.select_dtypes(include=[np.number]).columns
    cols_modificadas = []
    
    for col in cols_numericas:
        skewness = df_transformed[col].skew()
        if abs(skewness) > umbral:
            df_transformed[col] = np.log1p(np.maximum(0, df_transformed[col]))
            cols_modificadas.append(col)
            
    return df_transformed, cols_modificadas
```

```python
df = pd.read_parquet("hf://datasets/atalaydenknalbant/video-games-dataset/videogames_data.parquet")
```

```python
df_transf = corregir_asimetria(df)[0]
df_transf = df_transf.drop(columns=['clip'])
df_transf.head(10)
```

### AutoML y Experimentación Rápida Low-Code (LazyPredict y FLAML)

Las bibliotecas modernas de AutoML Low-Code permiten automatizar el cribado inicial de algoritmos (benchmarking) y la búsqueda de hiperparámetros manteniendo total compatibilidad con la API estándar de scikit-learn (.fit(), .predict()).

### Cribado Rápido de Modelos Base con LazyPredict

LazyPredict permite entrenar y evaluar decenas de algoritmos estándar de forma simultánea con apenas tres líneas de código. Es ideal como punto de partida (baseline) para identificar qué familias de modelos (lineales, árboles, ensambles, SVM) se adaptan mejor a la estructura de los datos antes de invertir tiempo en optimización.

```python
# 1. Inicialización del evaluador masivo (sin ajuste de hiperparámetros)
clf = LazyClassifier(verbose=0, ignore_warnings=True, custom_metric=None)

# 2. Entrenamiento y evaluación paralela de algoritmos sobre los splits

df_numerico = df_transf.select_dtypes(include=["int", "float"])

dfw = df_numerico.dropna(subset=['rating_top'])

X = dfw.drop(columns=['rating_top', 'rating'])
y = dfw['rating_top'].values.ravel()


# Supongamos X (matriz de características) e y (variable objetivo)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.2, 
    stratify=y,       # Conserva la proporción de clases
    random_state=42
)

modelos, predicciones = clf.fit(X_train, X_test, y_train, y_test)

# 3. Inspección de la tabla comparativa ordenada por rendimiento
print(modelos[["Accuracy", "Balanced Accuracy", "ROC AUC", "F1 Score"]].head(20))
```

### Optimización Ligera bajo Presupuesto de Tiempo con FLAML

FLAML (Fast and Lightweight AutoML, desarrollado por Microsoft) automatiza tanto la selección del estimador como la optimización de sus hiperparámetros basándose en un presupuesto de tiempo (time_budget), priorizando configuraciones de bajo costo computacional antes de explorar modelos complejos.

```python
# 1. Inicialización del motor AutoML
automl = AutoML()

# 2. Búsqueda automática de modelo e hiperparámetros en 30 segundos
automl.fit(
    X_train=X_train,
    y_train=y_train,
    task="classification",   # Soporta "classification", "regression", "ts_forecast"
    metric="log_loss",
    time_budget=30,
    seed=42
)

# 3. Inspección del estimador ganador y generación de predicciones
print(f"Mejor modelo encontrado: {automl.best_estimator}")
print(f"Configuración óptima:    {automl.best_config}")

y_pred = automl.predict(X_test)
print(f"F1-Score en Test:        {f1_score(y_test, y_pred, average='weighted'):.4f}")
```

### El Paradigma del Vibe Coding

El término Vibe Coding (acuñado originalmente por Andrej Karpathy) describe un flujo de desarrollo en el que el programador guía a la IA expresando deseos, correcciones e intenciones en lenguaje natural, confiando en que el agente mantenga la sintaxis, las dependencias y la compilación.

#### Cambio de Rol: Del "Escribano" al "Director de Orquesta"

- Desarrollo Tradicional: $80\%$ del tiempo escribiendo sintaxis, $20\%$ pensando en diseño y validación.
- Vibe Coding: $10\%$ especificando contexto e intenciones, $20\%$ revisando diffs de código, $70\%$ diseñando arquitectura, casos límite (edge cases) y pruebas de seguridad.

### Comparativa del Ecosistema de Herramientas Agénticas

|Herramienta|Tipo de Integración|Manejo de Contexto Global|Autonomía de Ejecución|Caso de Uso Ideal|
|-----------|-------------------|-------------------------|----------------------|-----------------|
|GitHub Copilot|Extensión IDE (VS Code, JetBrains)|Archivos abiertos y contexto local inmediato|Baja (Completado de código y chat inline)|Autocompletado rápido línea a línea y funciones cortas.|
|Cursor|IDE independiente (Fork de VS Code)|Indexación del repositorio completo mediante embeddings (@codebase)|Media-Alta (Edición multi-archivo con Composer)|Prototipado masivo, refactorizaciones globales y proyectos completos.|
|Windsurf|IDE independiente (Cascade Engine)|Flujos profundos de atención sobre el árbol de archivos|Alta (Colaboración simbiótica en tiempo real)|Mantenimiento continuo con comprensión profunda de estado.|
|Cline / Claude Dev|Extensión agéntica open-source|Contexto dinámico solicitado por herramienta (Tool Use)|Muy Alta (Ejecuta comandos de consola, lee/crea archivos y prueba)|Tareas complejas desatendidas: corrección de bugs, tests automatizados.|

### Flujos de Trabajo Human-in-the-Loop (HITL)

A pesar de la alta autonomía de los agentes, la supervisión humana es indispensable para evitar el sesgo de confirmación y la acumulación de deuda técnica.

#### Protocolo de Validación

1. Especificación Acotada (Scopeing): No solicitar "crea la app entera". Definir módulos atómicos con entradas y salidas estrictamente tipadas.

2. Inspección Visual de Diffs: Antes de aceptar un cambio (Accept Diff), verificar:

    - ¿Ha introducido dependencias innecesarias?
    - ¿Ha modificado archivos fuera del alcance solicitado?


3. Ejecución de Tests Automatizados: El agente debe escribir los tests unitarios (pytest) antes o simultáneamente a la implementación del módulo.

4. Refactorización Limpia: Solicitar explícitamente al agente reducir duplicidades y eliminar código sobrante (boilerplate).

```python
# Ejemplo de prompt de auditoría dentro del flujo HITL:
"""
PROMPT DE VALIDACIÓN Y CLEAN CODE:
"Revisa el módulo 'clustering_service.py' generado en el paso anterior.
1. Asegúrate de que todas las funciones tengan Type Hints explícitos.
2. Mueve la lógica de escalado a un Pipeline de Scikit-Learn reutilizable.
3. Añade docstrings formateados en estilo NumPy.
4. Escribe un test unitario en pytest que compruebe que DBSCAN maneja correctamente una matriz vacía."
"""
```

### Ideas Clave

- El peligro del "Aceptar Todo" (Blind Accept): Aceptar sugerencias de IA sin revisar el diff genera una acumulación acelerada de código muerto, vulnerabilidades de seguridad y pérdida del conocimiento sobre la base del proyecto.
- Instrucciones de Proyecto (.cursorrules / .clinerules): Configurar archivos de reglas en la raíz del proyecto para forzar al agente a cumplir estándares de estilo (ej. "Usa siempre type hints", "No uses bibliotecas externas sin consultar", "Sigue las convenciones de Scikit-Learn").
- Agentes con ejecución en terminal: Herramientas como Cline o Cursor Composer pueden ejecutar comandos (pytest, python script.py), leer la traza de error en consola y arreglar el código automáticamente en un bucle cerrado hasta que los tests pasen a verde.

### Alucinaciones en Código: Sintácticas, Semánticas y Lógicas

Aunque los modelos de lenguaje modernos tienen una tasa de sintaxis correcta muy elevada, presentan fallos sutiles pero críticos en entornos de producción:

### Ejemplos Típicos de Alucinación en Ciencia de Datos

1. Atributos inexistentes en pipelines: Intentar extraer pipeline.coef_ directamente en lugar de pipeline.named_steps['regressor'].coef_.
2. Métodos obsoletos: Utilizar argumentos deprecados como squared=False en mean_squared_error en versiones recientes de Scikit-Learn.
3. Imputación sesgada: Aplicar imputaciones globales sobre todo el conjunto de datos $X$ antes de realizar el split de entrenamiento y prueba (Data Leakage silencioso).

### Seguridad en el Desarrollo Asistido (OWASP for LLMs)

El desarrollo mediante asistencias de IA e interfaces conversacionales introduce vectores de ataque específicos que deben auditarse:

```python
# Ejemplo de vulnerabilidad por inyección de código/comandos (Code Injection)
# Código ingenuamente generado por un agente de IA para evaluar expresiones del usuario:

def evaluar_formula(expresion_usuario: str, df: pd.DataFrame):
    # VULNERABILIDAD CRÍTICA: La IA generó un eval() directo
    # Si un atacante envía: "__import__('os').system('rm -rf /')"
    return eval(expresion_usuario)

# Solución segura (Gobernanza / Code Review):
def evaluar_formula_segura(columna: str, df: pd.DataFrame):
    if columna in df.columns:
        return df[columna].describe()
    raise ValueError("Columna no permitida o inexistente.")
```

### Principales Vulnerabilidades a Auditar

- Inyección de Prompts (Prompt Injection): Manipulación de instrucciones para omitir reglas de seguridad del agente.
- Fuga de Datos Sensibles (Data Leakage / Telemetry): Envío inadvertido de claves API, tokens de conexión a bases de datos o datos regulados (GDPR, HIPAA) a los servidores de proveedores de IA durante el autocompletado.
- Alucinación de Paquetes (Package Hallucination / Typosquatting): El agente sugiere instalar una librería inexistente (ej. pip install sklearn-advanced-utils). Atacantes pueden registrar esos nombres en PyPI con malware embebido.

### Licencias de Código, Propiedad Intelectual y Privacidad

|Aspecto|Riesgo Asociado|Buena Práctica / Mitigación|
|-------|---------------|---------------------------|

### Marco de Gobernanza: El Principio de Confianza Cero (Zero Trust Code)

Para operar de forma segura bajo el paradigma de Vibe Coding, se debe aplicar el principio: "No confíes en el código generado; verifícalo siempre".

#### Reglas de Gobernanza para el Equipo de Desarrollo

1. Integración Continua (CI) Obligatoria: Ningún código aceptado por el agente puede unirse a la rama principal (main) sin pasar linters automáticos (ruff, mypy) y suite de tests.
2. Cobertura de Pruebas Mínima: Exigir un mínimo del $80\%$ de cobertura de pruebas unitarias sobre módulos generados por agentes.
3. Escaneo Estático de Seguridad (SAST): Ejecutar herramientas como Bandit o SonarQube para detectar vulnerabilidades en scripts Python generados por IA.

### Ideas Clave

- Responsabilidad final humana: Las alucinaciones no son excusa ante un fallo en producción. El desarrollador o científico de datos es el único responsable legal y técnico del código que firma en la revisión.
- Atención al Data Leakage en IA: Nunca pases variables de test o matrices de datos reales dentro de las instrucciones o prompts de un LLM comercial sin anonimizar o asegurar un canal privado.
- Revisión activa de dependencias: Antes de ejecutar pip install sugerido por un agente de Vibe Coding, verifica la existencia oficial de la biblioteca en PyPI y su número de descargas para evitar ataques de Typosquatting.
