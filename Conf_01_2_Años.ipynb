```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
import warnings
warnings.simplefilter(action='ignore', category=FutureWarning)
warnings.filterwarnings("ignore")
```

# **Ejercicio 1.**

Desarrolle una aplicación que genere un dataframe con todos los años de nuestra era (hasta el año 3000) y que, para cada uno de ellos indique:
a) el día de comienzo del año,
b) si el año es bisiesto o no y,
c) si es año secular o no.

```python
df = pd.DataFrame()
result = []

for i in range(1, 3001):

    if i % 400 == 0:
        secular = True
        bisiesto = True
    elif i % 100 == 0:
        secular = False
        bisiesto = False
    elif i % 4 == 0:
        secular = False
        bisiesto = True
    else:
        secular = False
        bisiesto = False

    if i % 100 == 0:
        mult100 = True
    else:
        mult100 = False

    if i % 4 ==0:
        mult4 = True
    else:
        mult4 = False

    a=np.remainder(1+5*np.remainder((i-1),4)+4*np.remainder((i-1),100)+6*np.remainder((i-1),400),7)
    if a == 0:
        dia = "domingo"
    elif a == 1:
        dia = "lunes"
    elif a == 2:
        dia = "martes"
    elif a == 3:
        dia = "miércoles"
    elif a == 4:
        dia = "jueves"
    elif a == 5:
        dia = "viernes"
    else:
        dia = "sábado"

    row = [i, secular, bisiesto, dia, mult100, mult4]
    result.append(row)

df = pd.DataFrame(result, columns=['Año', 'Secular', 'Bisiesto', 'Comienza en', 'Múltiplo_100', 'Múltiplo_4'])
df = df.reindex(columns=['Año', 'Comienza en', 'Bisiesto', 'Secular', 'Múltiplo_100', 'Múltiplo_4'])

df
```

# **Ejercicio 2.**

Entrene un modelo de Regresión Logística que permita predecir si un año será bisiesto o no a partir de los datos de Dataframe generado en el ejercicio anterior.

```python
X = df[['Año', 'Secular', 'Múltiplo_100', 'Múltiplo_4']]
y = df[['Bisiesto']]

print(X)
print(y)

x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)

logisticregression = LogisticRegression()

logisticregression.fit(x_train, y_train.values.ravel())

predictions = logisticregression.predict(x_test)
score = logisticregression.score(x_test, y_test)
params = logisticregression.get_params(deep=True)

print(score)
```

```python
prediccion = logisticregression.predict(X)
df['Predicción'] = pd.DataFrame(prediccion)
df = df.reindex(columns=['Año', 'Comienza en', 'Secular', 'Múltiplo_100', 'Múltiplo_4', 'Bisiesto', 'Predicción'])
df
```
